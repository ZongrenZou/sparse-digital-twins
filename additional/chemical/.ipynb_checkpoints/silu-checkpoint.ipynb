```python
# train_cfm_conditional_flax.py
# pip install jax jaxlib flax optax
from __future__ import annotations
from dataclasses import dataclass
from typing import Tuple
import os, json
from dataclasses import asdict, is_dataclass
from pathlib import Path
from typing import Any, Dict, Tuple

import numpy as np
from flax import serialization

import jax
import jax.numpy as jnp
import jax.random as jr
import optax
from flax import linen as nn
from flax.training import train_state
import scipy.io as sio
import numpy as np
import diffrax as dfx

from ckpt_io import save_model, load_model
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt


from typing import Tuple


import jax.numpy as jnp
from flax import linen as nn
from flax.training import train_state



class VelocityMLP(nn.Module):
    """v_theta(t, z, x): concat [z, t, x] -> MLP -> R^d."""
    hidden_dims: Tuple[int, ...]
    dim_out: int

    @nn.compact
    def __call__(self, t, z, x):
        # t: () or (B,), z: (d,) or (B,d), x: (k,) or (B,k)
        # Broadcast t to match batch of z/x if needed
        if t.ndim == 0:
            t = t[None] if z.ndim == 1 else jnp.broadcast_to(t, (z.shape[0],))
        t = t[..., None]  # (B,1) or (1,1)

        # Ensure rank alignment
        if z.ndim == 1:
            z = z[None, :]  # (1,d)
        if x.ndim == 1:
            x = x[None, :]  # (1,k)

        h = jnp.concatenate([z, t, x], axis=-1)
        for width in self.hidden_dims:
            # h = nn.tanh(nn.Dense(width)(h))
            h = nn.silu(nn.Dense(width)(h))
        out = nn.Dense(self.dim_out)(h)  # (B, d)
        return out if out.shape[0] > 1 else out[0]  # return (d,) if single


@dataclass
class SolverCfg:
    rtol: float = 1e-5
    atol: float = 1e-5
    max_steps: int = 1_000_000

def sample(
    key,
    params,
    x,
):
    # Build once
    term = dfx.ODETerm(lambda t, y, args: velocity(params, model, t, y, args))

    def solve_single(z0i, x):
        sol = dfx.diffeqsolve(
            term, dfx.Tsit5(),
            t0=0.0, t1=1.0, dt0=None,
            y0=z0i,
            args=x,                             # pass conditioning vector for this traj
            saveat=dfx.SaveAt(t1=True),
            stepsize_controller=dfx.PIDController(rtol=1e-5, atol=1e-5),
            max_steps=1_000_000,
        )
        return sol.ys

    # If you want n samples for the SAME x:
    # z0 = jr.normal(key, [x.shape[0], 4])
    z0 = np.random.normal(size=[x.shape[0], 4])
    z0 = jnp.array(z0)
    x_fixed = x  # (k,)
    zT = jax.vmap(solve_single, in_axes=(0, 0))(z0, x_fixed)
    return zT, z0


# ============== Inference helpers (plug these into your sampler) ==============

def velocity(params, model, t, z, x):
    """Single-trajectory: t:(), z:(d,), x:(k,) -> (d,)"""
    return model.apply({"params": params}, t, z, x)
    # return model.apply(params, t, z, x)
```

```python
M = 10000

params, model, extras = load_model("./checkpoints/cfm_ssa_v1_silu", VelocityMLP)
x_mu = extras["x_mu"]
x_sd = extras["x_sd"]
y_mu = extras["y_mu"]
y_sd = extras["y_sd"]


M = 10_000
data = sio.loadmat("./data/data_ssa_v1.mat")
_x_data = data["Xs"]
_dx_data = data["dXs"]
dt = data["tau"]

x_data = _x_data.reshape([-1, 4])
dx_data = _dx_data.reshape([-1, 4])
# seed = np.random.randint(0, 1e9)
seed = 58833216
np.random.seed(seed)
print(seed)
idx = np.random.choice(x_data.shape[0], x_data.shape[0], replace=False)[:M]
x_data = x_data[idx, :]
dx_data = dx_data[idx, :]


x = x_data[3:4, :]
# print(x)
# x = np.array([20, 70, 80, 250]).reshape([1, -1])
x = np.tile(x, [1000, 1])
key = jr.PRNGKey(72819)
z_samples, z0_samples = sample(
    key, params,
    x=(x - x_mu) / x_sd,
)
z_samples = z_samples[:, 0, :] * y_sd + y_mu
z_samples = np.array(z_samples)[..., 0]

c1 = 0.0016603021749958492
c2 = 0.0001
c3 = 0.1


#### dX1
P1 = np.random.poisson(c1 * x[0, 0:1] * x[0, 1:2] * dt, size=[1000, 1])
P2 = np.random.poisson(c2 * x[0, 2:3] * dt, size=[1000, 1])
z_samples_ref = -P1 + P2
# print(c1 * x[0, 0:1] * x[0, 1:2] * dt, c2 * x[0, 2:3] * dt)
# print(x[0, 0:1] * x[0, 1:2], x[0, 2:3])
print(x[0, 0:1], x[0, 1:2])
print("Reference values:", -c1 * x[0, 0:1] * x[0, 1:2] * dt + c2 * x[0, 2:3] * dt, c1 * x[0, 0:1] * x[0, 1:2] * dt + c2 * x[0, 2:3] * dt)
print("Flow matching:", np.mean(z_samples), np.std(z_samples)**2)
print("Reference sampler:", np.mean(z_samples_ref), np.std(z_samples_ref)**2)

# #### dX4
# z_samples_ref = np.random.poisson(c3 * x[0, 2:3] * dt, size=[1000, 1])
# print("Reference values:", c3 * x[0, 2:3] * dt, c3 * x[0, 2:3] * dt)
# print("Flow matching:", np.mean(z_samples), np.std(z_samples)**2)
# print("Reference sampler:", np.mean(z_samples_ref), np.std(z_samples_ref)**2)


plt.figure()
plt.hist(z_samples, bins=30, density=True)
plt.title("Samples from the flow matching model with x being {}.".format(str(x[0])))
plt.savefig("Samples from the trained flow matching model")
plt.figure()
plt.hist(z_samples_ref, bins=30, density=True)
plt.title("Samples from the exact generative model with x being {}.".format(str(x[0])))
plt.savefig("Samples from the exact generative model")

print("End main.")
```
