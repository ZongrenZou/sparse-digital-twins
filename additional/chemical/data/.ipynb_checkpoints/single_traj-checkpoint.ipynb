```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.io as sio
```

```python
def tau_leap_mm(
    tfinal=50.0,
    tau=0.2,                 # like the step used for CLE demo in the paper
    seed=100,
):
    """
    Tau-leaping for the Michaelis–Menten system in Higham (SIAM Review, 2008).
    Reactions:
      (1) S1 + S2 --c1--> S3
      (2) S3       --c2--> S1 + S2
      (3) S3       --c3--> S4 + S2

    Update (paper eq. (11)):
      X(t+tau) = X(t) + sum_j nu_j * Poisson(a_j(X(t)) * tau)
    """
    rng = np.random.default_rng(seed)

    # ---------- Stoichiometry (columns are ν1, ν2, ν3), as in Listing 1 ----------
    V = np.array([
        [-1, +1,  0],   # S1
        [-1, +1, +1],   # S2
        [+1, -1, -1],   # S3
        [ 0,  0, +1],   # S4
    ], dtype=int)

    # ---------- Parameters & initial conditions from Section 8 ----------
    nA  = 6.023e23      # Avogadro
    vol = 1e-15         # liters
    # rate constants
    k1, k2, k3 = 1e6, 1e-4, 1e-1
    # stochastic rates (paper mapping): c1 = k1/(nA*vol), c2 = k2, c3 = k3
    c1 = k1 / (nA * vol)
    c2 = k2
    c3 = k3

    # initial concentrations (M) -> counts by rounding
    X1_0 = int(round(5e-7 * nA * vol))
    X2_0 = int(round(2e-7 * nA * vol))
    X = np.array([X1_0, X2_0, 0, 0], dtype=int)   # [S1,S2,S3,S4]

    # time grid
    n_steps = int(np.ceil(tfinal / tau))
    t = np.linspace(0.0, n_steps * tau, n_steps + 1)
    X_traj = np.empty((n_steps + 1, 4), dtype=int)
    X_traj[0] = X

    for k in range(1, n_steps + 1):
        # ---------- Propensities a_j(X) ----------
        a1 = c1 * X[0] * X[1]     # bimolecular
        a2 = c2 * X[2]            # unimolecular
        a3 = c3 * X[2]            # unimolecular
        a = np.array([a1, a2, a3], dtype=float)

        # ---------- Poisson firings over [t, t+tau) (paper eq. (11)) ----------
        lam = np.maximum(a * tau, 0.0)
        firings = rng.poisson(lam)

        # ---------- Simple nonnegativity safeguard (optional) ----------
        # (1) cannot fire more than min(S1,S2); (2),(3) cannot fire more than S3.
        # This is a light "post-leap" check; decrease tau for less bias.
        max_f = np.array([min(X[0], X[1]), X[2], X[2]])
        firings = np.minimum(firings, max_f)

        # ---------- State update ----------
        X = X + V @ firings
        X = np.maximum(X, 0)
        X_traj[k] = X

    return t, X_traj
```

```python
t, X = tau_leap_mm(tfinal=50.0, tau=0.2, seed=123)
```

```python
plt.plot(t, X[:, 0])
plt.plot(t, X[:, 1])
plt.plot(t, X[:, 2])
plt.plot(t, X[:, 3])
plt.legend(["X_1", "X_2", "X_3", "X_4"])
plt.xlabel("t")
plt.ylabel("X")
plt.show()
```
