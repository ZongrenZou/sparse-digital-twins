```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
ratios_1 = np.loadtxt("./outputs/ratios_1.txt")
ratios_2 = np.loadtxt("./outputs/ratios_2.txt")
ratios_3 = np.loadtxt("./outputs/ratios_3.txt")
```

```python
# we plot the results
fig, ax = plt.subplots(1, 2, figsize=[12, 5], dpi=300)
# plt.suptitle("Finding ancestors of " + target_name)
ax[0].plot(["3", "2", "1", "0"], ratios_1, "k-o", markersize=10)
ax[0].plot(["3", "2", "1", "0"], ratios_2, "b--^", markersize=10)
ax[0].plot(["3", "2", "1", "0"], ratios_3, "r-s", markersize=10)
ax[0].set_xlabel("Number of ancestors", fontsize=15)
ax[0].legend(["$\Delta x_{1}$", "$\Delta x_{2}$", "$\Delta x_{3}$"], fontsize=15)
ax[0].set_title("The noise-to-signal ratio", fontsize=15)

# ax[1].plot(["$x_2$", "$x_1$", "$x_3$"], np.array(ratios[1:]) - np.array(ratios[:-1]), "k-o")
increments_1 = np.array(ratios_1[1:]) - np.array(ratios_1[:-1])
increments_2 = np.array(ratios_2[1:]) - np.array(ratios_2[:-1])
increments_3 = np.array(ratios_3[1:]) - np.array(ratios_3[:-1])
ax[1].plot(["1", "2", "3"], increments_1, "k-o", markersize=10)
ax[1].plot(["1", "2", "3"], increments_2, "b--^", markersize=10)
ax[1].plot(["1", "2", "3"], increments_3, "r-s", markersize=10)
# ax[1].annotate("$x_2$", (0, increments_1[0]), (0.05, increments_1[0]), fontsize=20, color="k")
# ax[1].annotate("$x_1$", (0, increments_1[1]), (1.05, increments_1[1]), fontsize=20, color="k")
# ax[1].set_xticks([])
# ax[1].annotate(["$x_2$", "$x_1$", "$x_3$"], (np.arange(1, 4), increments_1))
ax[1].set_title("The increment of the noise-to-signal ratio", fontsize=15)
ax[1].set_xlabel("Number of ancestors being pruned", fontsize=15)
ax[1].legend(["$\Delta x_{1}$", "$\Delta x_{2}$", "$\Delta x_{3}$"], fontsize=15, loc=2)
# ax[1].set_xlabel("Pruned variables")
plt.tight_layout()
fig.savefig("./figs/ratios.png", dpi=200)
```
