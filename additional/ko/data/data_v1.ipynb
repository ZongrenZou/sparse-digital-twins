```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.io as sio
```

\begin{gather*}
dx_1 = x_1x_3dt + \sigma_1dW_1,\\
dx_2 = -x_2x_3dt + \sigma_2dW_2,\\
dx_3 = (-x_1^2 + x_2^2) dt + \sigma_3 dW_3.
\end{gather*}

```python
def simulate(T, dt, sigma, y0s):
    N = int(T / dt)
    print(N)
    M = y0s.shape[-1]
    y = np.zeros([3, N + 1, M])
    dy = np.zeros([3, N + 1, M])
    y[:, 0:1, :] = y0s.reshape([3, 1, -1])  # initial conditions

    # Generate Brownian increments
    dW = np.random.normal(0, 1, size=[3, N, M])
    dW = np.sqrt(dt) * dW

    # Euler-Maruyama method
    for i in range(N):
        y1, y2, y3 = y[:, i, :]
        dy1 = y1 * y3 * dt + 1 * sigma[0] * dW[0, i, :]
        print(np.mean(np.abs(y1 * y3 * dt)), np.mean(np.abs(sigma[0] * dW[0, i, :])))
        dy2 = -y2 * y3 * dt + sigma[1] * dW[1, i, :]
        dy3 = (-(y1**2) + y2**2) * dt + sigma[2] * dW[2, i, :]
        dy[:, i, :] = np.array([dy1, dy2, dy3])
        y[:, i + 1, :] = y[:, i, :] + dy[:, i, :]

    return y, dy
```

```python
# np.random.seed(5910)
np.random.seed(8762)
T = 10.0
dt = 0.001
sigma = 3 * [0.1]
M = 100
y0s = 2 * np.random.normal(size=[3, M])

ys, _ = simulate(T, dt, sigma, y0s)
```

```python
x_data = ys[:, :, :]
dx_data = ys[:, 1:, :] - ys[:, :-1, :]
x_data = ys[:, :-1, :]
```

```python
x_data_train = x_data[:, :, :50]
x_data_test = x_data[:, :, 50:]
dx_data_train = dx_data[:, :, :50]
dx_data_test = dx_data[:, :, 50:]
```

```python
sio.savemat(
    "./data.mat",
    {
        "x_data_train": x_data_train,
        "x_data_test": x_data_test,
        "dx_data_train": dx_data_train,
        "dx_data_test": dx_data_test,
    },
)
```
