```python
from __future__ import annotations
from dataclasses import dataclass
from typing import Tuple
import os, json
from dataclasses import asdict, is_dataclass
from pathlib import Path
from typing import Any, Dict, Tuple

import numpy as np
from flax import serialization

import jax
import jax.numpy as jnp
import jax.random as jr
import optax
from flax import linen as nn
from flax.training import train_state
import scipy.io as sio
import numpy as np
import diffrax as dfx

from ckpt_io import save_model, load_model
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

import models


@dataclass
class SolverCfg:
    rtol: float = 1e-5
    atol: float = 1e-5
    max_steps: int = 1_000_000

def sample(
    key,
    params,
    x,
    z0=None,
    M=None,
):
    # Build once
    term = dfx.ODETerm(lambda t, y, args: velocity(params, model, t, y, args))

    @jax.jit
    def solve_single(z0i, x):
        sol = dfx.diffeqsolve(
            term, dfx.Tsit5(),
            t0=0.0, t1=1.0, dt0=None,
            y0=z0i,
            args=x,                             # pass conditioning vector for this traj
            saveat=dfx.SaveAt(t1=True),
            stepsize_controller=dfx.PIDController(rtol=1e-5, atol=1e-5),
            max_steps=1_000_000,
        )
        return sol.ys

    if z0 is None:
        z0 = np.random.normal(size=[x.shape[0], 1])
        z0 = jnp.array(z0)
    x_fixed = x 
    zT = jax.vmap(solve_single, in_axes=(0, 0))(z0, x_fixed)
    return zT, z0


# ============== Inference helpers (plug these into your sampler) ==============

def velocity(params, model, t, z, x):
    """Single-trajectory: t:(), z:(d,), x:(k,) -> (d,)"""
    return model.apply({"params": params}, t, z, x)
    # return model.apply(params, t, z, x)
```

```python
data = sio.loadmat("./data/data_m_10.mat")
trajs_train = data["trajs_train"]
trajs_test = data["trajs_test"]

m = data["m"].reshape([]).astype(np.int32)
```

#### After prune

```python
params, model, extras = load_model("./checkpoints/cfm_m_10_pruned_3", models.VelocityMLP)
x_mu = extras["x_mu"].reshape([1, -1])
x_sd = extras["x_sd"].reshape([1, -1])
y_mu = extras["y_mu"].reshape([1, -1])
y_sd = extras["y_sd"].reshape([1, -1])
```

```python
def simulate(idx, M=1, N=101):
    # init = trajs_train[idx, :m].reshape([1, m])
    init = trajs_test[:M, :m].reshape([-1, m])
    us = [
        init,
    ]
    predictions = [us[-1][:, -1:]]
    for i in range(N-m):
        u_window = us[-1]
        update, _ = sample(
            None, 
            params, 
            x=(u_window[:, 7:10]-x_mu)/x_sd, 
            z0=None, 
            M=M,
        )
        u_new = u_window[:, -1:] + update[:, 0, :] * y_sd + y_mu
        new_u_window = jnp.concatenate([u_window[:, 1:], u_new], axis=-1)
        us += [new_u_window]
        predictions += [u_new]
    return np.stack(predictions, axis=0)
```

```python
N = 5001 + m
M = 5_000
predictions_2 = simulate(0, M=M, N=N)[..., 0].T
references = trajs_test[:M, m-1:N]
```

```python
def integrand(x,U0,x0,beta,sigma):
    v = np.exp(-beta * (U0 * (1 - (x/x0)**2)**2 + sigma * x / (2*x0)))        # integrand function
    return v 


gamma = 1                   # Coupling constant 
beta = 1                    # Inverse temperature, i.e. T=1/beta
U0 = 4                      # Depth of the well 
x0 = 1                      # Basin of the well
sigma = 0                   # Asymmetry of the well 


from scipy.integrate import quad


I = quad(integrand, -10, 10, args=(U0,x0,beta,sigma))

x = np.linspace(-2, 2, 1001)
pdf = np.exp(-beta * (U0 * (1- (x/x0)**2)**2 + sigma * x / (2 * x0)))/I[0]
```

```python
plt.plot(x, pdf, "k-")
plt.hist(predictions_2[:, :].flatten(), bins=500, density=True, histtype="step", linestyle="-", color="r")
plt.show()
```
