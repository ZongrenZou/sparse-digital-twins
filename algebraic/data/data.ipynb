```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.io as sio
from scipy.stats import kurtosis
```

#### Generate training data

```python
N = 10_000
x1 = -0.5 + 1 * np.random.uniform(size=[N, 1])
omega = np.random.normal(size=[N, 1])
y1 = x1 * omega

y2 = -1 + 2 * np.random.uniform(size=[N, 1])
x3 = y2**2

x2 = np.random.uniform(size=[N, 1])
xi = np.random.normal(size=[N, 1])
eta = 1 / np.sqrt(2) * np.random.laplace(size=[N, 1])
y3 = np.sqrt(1 - np.sin(np.pi * x2)) * xi + np.sqrt(np.sin(np.pi * x2)) * eta
```

```python
plt.figure(figsize=[3, 2])
plt.plot(x1, y1, "x")
plt.show()
```

```python
plt.figure(figsize=[3, 2])
plt.plot(x3, y2, "x")
plt.show()
```

```python
plt.figure(figsize=[3, 2])
plt.plot(x2, y3, "x")
plt.show()
```

```python
X = np.concatenate(
    [x1, x2, x3],
    axis=-1,
)
Y = np.concatenate(
    [y1, y2, y3],
    axis=-1,
)

sio.savemat(
    "./data_train.mat",
    {
        "x_data": X,
        "y_data": Y,
    },
)
```

#### Generate test data (on which we evaluate our generative models)

```python
N = 500
x1 = -0.5 + 1 * np.random.uniform(size=[N, 1])
x2 = np.random.uniform(size=[N, 1])

y2 = -1 + 2 * np.random.uniform(size=[N, 1])
x3 = y2**2

X = np.concatenate(
    [x1, x2, x3],
    axis=-1,
)
sio.savemat("./data_test.mat", {"x_data": X})
```
